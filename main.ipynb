# Berlin Electoral Districts geodata - Sample Notebook

## Explore data

```python

import folium
import json
from IPython.display import display

# Read geojson file
with open(
    file="./data/02-silver/berlin-electoral-districts-federal-election-2025/berlin-electoral-districts-federal-election-2025.geojson", mode="r", encoding="utf-8"
) as geojson_file:
    geojson = json.load(geojson_file, strict=False)
            
```

```python

# Create a folium map
m = folium.Map(location=[52.516388888889, 13.377777777778], zoom_start=10)
            
```

```python

# Add the GeoJSON layer to the map
folium.GeoJson(
    geojson,
    name='GeoJSON Layer',
    style_function=lambda x: {
        'fillColor': 'blue',
        'color': 'black',
        'weight': 1,
        'fillOpacity': 0.5
    }
).add_to(m)
            
```
